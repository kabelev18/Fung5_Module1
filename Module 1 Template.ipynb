## Disease Background: 

* Prevalence & incidence

Alzheimer’s disease is the most common cause of dementia and its prevalence increases significantly with age. In 2026, an estimated 7.4 million Americans age 65 and older are living with clinical Alzheimer’s dementia, which is approximately 1 in 9 adults over age 65. About 74% of these individuals are 75 years or older. Prevalence increases from approximately 5.2% among adults ages 65–74 to 35.8% among adults age 85 and older. The number of Americans age 65 and older living with Alzheimer’s is projected to increase to approximately 13.8 million by 2060 if major advances in prevention or treatment do not occur. Incidence also increases strongly with age. Earlier U.S. population estimates found annual incidence rates of approximately 0.4% for ages 65–74, 3.2% for ages 75–84, and 7.6% for those age 85 and older (1).

* Economic burden

Alzheimer’s disease creates a large economic burden through medical care, long-term care, hospice services, and unpaid caregiving. In the United States, health care, long-term care, and hospice costs for people age 65 and older with Alzheimer’s or other dementias are projected to reach approximately $409 billion in 2026. Medicare and Medicaid are expected to cover about $263 billion of these costs, while approximately $103 billion will come from out-of-pocket spending. The economic burden extends beyond direct medical expenses because more than 12 million family members and other unpaid caregivers provided approximately 19.6 billion hours of dementia care in 2025, valued at approximately $446.3 billion. The total lifetime cost of care for an individual with dementia has also been estimated at over $400,000, with families carrying a substantial portion through unpaid care and personal expenses (1).

* Risk factors (genetic, lifestyle) 

The development of Alzheimer’s disease is influenced by a combination of age, genetics, health conditions, lifestyle, and environmental factors. Age is the greatest known risk factor, with most people developing Alzheimer’s after age 65. Family history also contributes to risk. One of the most important genetic risk factors is the APOE ε4 allele, which increases the likelihood of developing Alzheimer’s and is associated with earlier disease onset in some populations. Rare mutations in APP, PSEN1, and PSEN2 can directly cause familial early-onset Alzheimer’s disease. However, these mutations account for only a small portion of Alzheimer’s cases, and carrying a risk gene such as APOE ε4 does not guarantee that an individual will develop the disease (2). Several potentially modifiable health and lifestyle factors are also associated with dementia risk. These include physical inactivity, high blood pressure, diabetes, obesity, smoking, excessive alcohol consumption, hearing loss, and poor cardiovascular health. Social isolation, inadequate sleep, and lack of mental stimulation may also contribute to risk. Therefore, managing cardiovascular and metabolic health and maintaining physical, cognitive, and social activity may contribute to better brain health, although these behaviors cannot guarantee prevention of Alzheimer’s disease (3).

* Societal determinants

Social determinants of health can influence Alzheimer’s and dementia risk by affecting an individual's education, socioeconomic status, access to health care, living environment, and social connections. Lower educational attainment has been associated with poorer cognitive health, potentially because education contributes to "cognitive reserve," allowing the brain to better compensate for age- and disease-related changes. Limited access to health care may also delay the diagnosis and treatment of conditions such as hypertension and diabetes that are associated with dementia risk. Additionally, unsafe neighborhoods, limited opportunities for exercise and healthy food, environmental exposures, and social isolation may negatively affect brain health throughout life (4). There are also important disparities in Alzheimer’s prevalence in the United States. Studies have reported higher prevalence among older Black and Hispanic adults compared with older White adults. Current evidence suggests these differences cannot be explained simply by genetics; differences in socioeconomic conditions, cardiovascular disease, diabetes, health care access, education, and other lifetime social and environmental exposures likely contribute to these disparities (1). 

* Symptoms

The symptoms of Alzheimer’s disease generally begin gradually and become more severe as neuronal damage progresses. Memory problems are commonly one of the earliest symptoms, particularly difficulty remembering recently learned information. Early symptoms can also include difficulty finding words, impaired reasoning or judgment, visual-spatial problems, repeatedly asking the same questions, losing objects, becoming lost in familiar locations, and difficulty completing normal tasks such as cooking or managing finances. As Alzheimer’s progresses, individuals may experience increasing confusion, personality or behavioral changes, agitation, and greater difficulty independently completing daily activities. In advanced disease, cognitive and physical function can become severely impaired, eventually affecting communication, mobility, swallowing, and the ability to independently perform basic activities (5).

* Diagnosis

Alzheimer’s disease is diagnosed using a combination of medical history, cognitive and neurological examinations, laboratory testing, brain imaging, and increasingly Alzheimer’s-specific biomarkers. Physicians may evaluate memory, attention, language, problem solving, and the patient's ability to perform normal daily activities while also speaking with family members about changes in behavior or personality. Blood and other laboratory tests can help rule out other causes of cognitive impairment. Structural brain imaging such as MRI or CT can identify brain atrophy and help exclude conditions such as tumors, strokes, bleeding, or fluid buildup (6). Alzheimer’s-specific biomarkers can provide stronger evidence of the underlying disease. Amyloid PET imaging can detect beta-amyloid plaque accumulation, while cerebrospinal fluid obtained through lumbar puncture can be analyzed for amyloid-beta, total tau, and phosphorylated tau. Blood-based biomarkers are also becoming increasingly important, particularly measurements involving phosphorylated tau such as p-tau217. Biomarker results are interpreted together with the patient's clinical evaluation rather than in isolation because a complete diagnosis requires considering both biological evidence and the patient's symptoms (7).


* Standard of care treatments (& reimbursement) 

The standard of care for Alzheimer’s disease depends on its stage and focuses mainly on slowing its progression and managing overall symptoms to better the patient’s quality of life as a full cure does not exist currently. Early-stage patients and those with only mild cognitive impairment can take lacenemab (Leqembi) and/or donanemab (Kisnula) which are both intravenous antibodies meant to reduce amyloid plaques and decrease cognitive decline. Other medications such as cholinesterase inhibitors (Donepezil, Rivastigmine, or Galantamine) can be taken at any stage to increase acetylcholine levels, Memantine (Namenda) for more severe stages can be used as a neurotransmitter antagonist, and Brexpiprazole (Rexulti) can help alleviate overall Alzheimer’s-caused agitation. In general, patients are recommended to take on physical activity and cognitive engagement to support overall health (8). For patients enrolled in Medicare with confirmed diagnosis of Alzheimer’s from PET scan or CSF data to support, Medicare-B covers 80% of eligible anti-amyloid treatments under its Coverage with Evidence Development policy, Medicare-D covers oral medications, and general Medicare can cover diagnostic MRIs, CT scans, PET scans, and care plan services under certain billing codes (hospice care not usually included)(9).

* Disease progression & prognosis

Alzheimer’s disease is typically diagnosed using a combination of medical history, cognitive, and neurological examinations (such as MMSE that tests attention, language, mathematical calculation, and memory), imaging, laboratory tests, and neuropsychological testing (upon necessity) to indicate identifying behaviors and abnormal brain function(10). Alzheimer’s generally progresses in three stages: mild, moderate, and severe, but the duration and symptoms of each stage can vary between patients. The mild stage is when patients experience mild memory loss and difficulties with planning, language, and other cognitive tasks but is otherwise relatively independent. The moderate stage is where patients exhibit deep memory loss, confusion, personality changes, and difficulty performing daily activities, often requiring assistance. Lastly, the severe stage is when patients experience profound cognitive and physical decline including loss of communication and mobility, difficulty swallowing, and require the need for constant care during all tasks(11).

* Continuum of care providers

Care for Alzheimer’s patients varies based on the patient’s stage of disease. During the early stage, patients can remain in their homes with visiting nurses, home health aides, and other companions to help with bathing, cooking, and medication management, or they can attend adult day-cares where they can be provided social and therapeutic activities. As patients progress into the middle stages of disease assisted living facilities and memory care units can provide structured routines, assistance with day-to-day tasks, housing, and capable staff. Once patients reach the late stage they usually are placed in End-of-Life Care facilities like nursing homes or hospice care facilities to provide 24/7 complex care, medical monitoring, treatment, and general comfort and pain management in their final stages of life. Continuing Care Retirement Communities (CCRCs) can also combine independent living, assisted living, skilled nursing, and memory care all into one location where patients can be cared for throughout the progression of the disease for smooth transitions through all stages(12).

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

Alzheimer’s disease is noted as a progressive build-up of amyloid-beta plaques outside neurons and tau protein tangles within neurons causing widespread synaptic dysfunction and brain degeneration. Anatomically, Alzheimer’s damage begins in the brain’s memory structures the hippocampus and entorhinal cortex before spreading through neural networks to the cerebral cortex and the nucleus basalis of Meynert (which sends acetylcholine to the cortex) leading to severe brain shrinkage. Physiologically, Alzheimer’s disrupts synaptic neurotransmission and plasticity (ability to adjust based on the body’s needs) and leads to overall damage in glucose metabolism in anatomically affected areas, leading to heavy cognitive decline. Notably, microglia and astrocytes in the brain become chronically activated as a result of the tau-tangles and AB plaques, which causes neuroinflammation and increased tissue damage(13). At the cellular and molecular level, abnormal processing of amyloid precursor protein produces toxic amyloid-beta peptides, while hyperphosphorylated tau detaches from microtubules and forms neurofibrillary tangles. These changes contribute to mitochondrial dysfunction, oxidative stress, neuronal damage, and ultimately lead to apoptotic cell death and progressive loss of brain function(14).

* Clinical Trials/next-gen therapies

Research into Alzheimer’s disease is expanding beyond the currently approved anti-amyloid therapies such as Leqembi and Kisunla with more oral, pill therapies on the horizon. Hundreds of clinical trials are investigating new treatments, including next-generation amyloid-targeting drugs, oral therapies, and approaches that target pathways other than amyloid. Researchers are also studying non-pharmacological approaches such as brain stimulation and focused ultrasound, as well as lifestyle interventions that may reduce the risk of developing Alzheimer’s. Another major area of research is treating patients before significant symptoms appear, as demonstrated by the AHEAD study that is investigating early intervention in people at high risk for Alzheimer’s. Together, these studies aim to expand treatment options and potentially delay or prevent cognitive decline before substantial brain damage occurs(15).

Comparison of tTAU Levels by Sex

To determine whether tTAU levels differed between female and male patients, the tTAU values were separated into two groups based on sex. The mean and standard deviation were calculated for each group. An independent two-sample t-test was then performed using Welch's t-test (equal_var=False) to compare the tTAU levels between the two groups.

```python
female_tTAU = []
male_tTAU = []

# sorting the patients by gender
for patient in Patient.all_patients:

    if patient.sex == "Female":
        female_tTAU.append(patient.tTAU)

    if patient.sex == "Male":
        male_tTAU.append(patient.tTAU)

# calculating mean tTAU levels in female and male patients
female_mean = statistics.mean(female_tTAU)
male_mean = statistics.mean(male_tTAU)

#calculating the standard deviation for female and male patients
female_stdev = statistics.stdev(female_tTAU)
male_stdev = statistics.stdev(male_tTAU)

# running a t-test on the data set to produce a p-value and t-statistic
t_stat, p_value = stats.ttest_ind (female_tTAU, male_tTAU, equal_var=False)

# printing results
print("Female tTAU Mean: ")
print("Mean =", female_mean)
print("Standard deviation =", female_stdev)
print("Male tTAU Mean: ")
print("Mean =", male_mean)
print("Standard deviation = ", male_stdev)

print("T-test results: ")
print("t-statistic = ", t_stat)
print("p-value =", p_value)
```

The mean tTAU level for female patients was 429.9 pg/µg, with a standard deviation of 323.6. The mean tTAU level for male patients was 603.8 pg/µg, with a standard deviation of 1184.8. The independent t-test produced a t-statistic of -0.8234 and a p-value of 0.4158. Because the p-value was 0.4158 which is much greater than 0.05, the results did not provide statistically significant evidence of a difference in tTAU levels between the two groups. Further, the very large standard deviations suggest insignificance when analyzing the average tTAU levels between the groups as there is excessive overlap in error bars as depicted in the graph below.

To visualize the difference in average tTAU levels between the two groups, a bar graph was generated showing the mean tTAU level for females and males. Standard deviation was included as an error bar to show the variation within each group. The t-test statistic and p-value were also displayed on the figure.

```python
# setting up the variables to be displayed in the bar graph
sex_labels = ["Female", "Male"]
mean_sex = [female_mean, male_mean]
stdev_sex = [female_stdev, male_stdev]

yerr = [np.zeros(len(mean_sex)), stdev_sex]

# plotting the bar graph with correct alignment of variables as well as printing the t and p-values.
plt.bar(sex_labels, mean_sex, yerr=yerr, capsize=10)

plt.text(0.5, max(mean_sex) + max(stdev_sex) - 100, "t = " + str(round(t_stat, 2)) + "\np = " + str(round(p_value, 4)), ha="center")

plt.title("Average tTAU Levels by Sex")
plt.xlabel("Sex")
plt.ylabel("Average tTAU Levels")
plt.show()
```

Figure 1. Average tTAU levels for female and male patients. Error bars represent standard deviation. The t-statistic and p-value from the independent t-test are displayed above the bars.

Relationship Between Age at Death and tTAU

To investigate whether age at death was associated with tTAU levels, a linear regression was performed separately for female and male patients. For the female group, age at death was used as the independent variable and tTAU level was used as the dependent variable. A scatter plot with a linear regression line was generated, and the slope, intercept, and R² value were calculated.

```python
# creating empty lists to fill the data
female_age = []
female_tTAU_values = []

# filtering out only female patients and their tTAU levels
for patient in Patient.all_patients:

    if patient.sex == "Female":
        female_age.append(patient.age_at_death)
        female_tTAU_values.append(patient.tTAU)

# setting up the scatter plot
female_x = np.array(female_age).reshape(-1, 1)
female_y = np.array(female_tTAU_values)

# running a linear regression
female_model = LinearRegression()
female_model.fit(female_x, female_y)

female_x_sorted = np.sort(female_x, axis=0)
female_y_pred = female_model.predict(female_x_sorted)

# getting the slope of the linear regression
female_slope = female_model.coef_[0]
female_intercept = female_model.intercept_
female_r2 = female_model.score(female_x, female_y)

# printing the linear regression results
print("Female regression results:")
print("Slope =", female_slope)
print("Intercept =", female_intercept)
print("R-squared =", female_r2)

# printing the scatter plot image with correct variable display and linear regression results visible
plt.scatter(female_age, female_tTAU_values)
plt.plot(female_x_sorted, female_y_pred)

plt.title("Female tTAU Levels vs. Age at Death")
plt.xlabel("Age at Death")
plt.ylabel("tTAU Levels")

plt.text(0.05, 0.90, "y = " + str(round(female_slope, 2)) + "x + " + str(round(female_intercept, 2)) + "\nR^2 = " + str(round(female_r2, 4)), transform=plt.gca().transAxes)

plt.show()
```

The female regression produced a slope of -1.331 and an R² value of 0.000921. The slope suggests a slightly negative correlation between the two variables (lower tTAU levels associated with later age at death) but the the R² value indicates that that linear regression is not a good fit for the data set because it sigifiicantly far away from 1, meaning that conclusion cannot be drawn. From a visual standpoint, the data does not showcase a clear trend as there are many outliers with either very high tTAU levels or low ages at death that do not correspond to a clear pattern.

For the male patients, the same linear regression procedure was used, with age at death as the independent variable and tTAU level as the dependent variable.

```python
# creating empty lists to fil for data collection
male_age = []
male_tTAU_values = []

# sorting out only male patients
for patient in Patient.all_patients:

    if patient.sex == "Male":
        male_age.append(patient.age_at_death)
        male_tTAU_values.append(patient.tTAU)

# running linear regression and setting up scatter plot of data (basically the same as for females in the cell above)
male_x = np.array(male_age).reshape(-1, 1)
male_y = np.array(male_tTAU_values)
male_x_sorted = np.sort(male_x, axis=0)
male_model = LinearRegression()
male_model.fit(male_x, male_y)

male_y_pred = male_model.predict(male_x_sorted)

male_slope = male_model.coef_[0]
male_intercept = male_model.intercept_
male_r2 = male_model.score(male_x, male_y)

# creating the scatter plot and printing all data and linear regression results
plt.scatter(male_age, male_tTAU_values)
plt.plot(male_x_sorted, male_y_pred)
plt.title("Male tTAU Levels vs. Age at Death")
plt.xlabel("Age at Death")
plt.ylabel("tTAU Levels")

plt.text(
    0.05,
    0.90,
    "y = " + str(round(male_slope, 2)) +
    "x + " + str(round(male_intercept, 2)) +
    "\nR² = " + str(round(male_r2, 4)),
    transform=plt.gca().transAxes
)

plt.show()

print("Male regression results:")
print("Slope =", male_slope)
print("Intercept =", male_intercept)
print("R-squared =", male_r2)
```

The male regression produced a slope of -16.543 and an R² value of 0.0156858. The slope suggests a slightly more negative correlation between the two variables than in the female sample, but again the the R² value indicates that that linear regression is not a good fit for the data set because it sigifiicantly far away from 1. It is clear from the plot that there is one outlier that is shifting the results of the regression.

Finally, a linear regression was performed using all patients to determine the overall relationship between age at death and tTAU levels, regardless of sex. This analysis used age at death as the independent variable and tTAU level as the dependent variable.

```python
# creating empty lists to fill with data
all_age = []
all_tTAU = []

# gathering tTAU levels from patients
for patient in Patient.all_patients:
    all_age.append(patient.age_at_death)
    all_tTAU.append(patient.tTAU)

all_x = np.array(all_age).reshape(-1, 1)
all_y = np.array(all_tTAU)

# running linear regression on the combined data
all_model = LinearRegression()
all_model.fit(all_x, all_y)

all_x_sorted = np.sort(all_x, axis=0)
all_y_pred = all_model.predict(all_x_sorted)

all_slope = all_model.coef_[0]
all_intercept = all_model.intercept_
all_r2 = all_model.score(all_x, all_y)

# printing results of linear regression
print("Overall regression results:")
print("Slope =", all_slope)
print("Intercept =", all_intercept)
print("R-squared =", all_r2)

# creating scatter plot of all data with linear regression results printed
plt.scatter(all_age, all_tTAU)
plt.plot(all_x_sorted, all_y_pred)
plt.title("tTAU Levels vs. Age at Death")
plt.xlabel("Age at Death")
plt.ylabel("tTAU Levels")
plt.text(
    0.05,
    0.90,
    "y = " + str(round(all_slope, 2)) +
    "x + " + str(round(all_intercept, 2)) +
    "\nR² = " + str(round(all_r2, 4)),
    transform=plt.gca().transAxes
)

plt.show()
```

Overall, the regression analysis shows that there is a negative relationship between age at death and tTAU levels. The regression line has a slope of -9.7081 and an R² value of 0.0099. Ultimately, this suggests that age at death has a relatively weak linear association with tTAU levels in the combined patient population as seen by the very low R^2 value 0.0099. This suggests that a conclusion cannot be drawn stating that patients with higher tTAU levels died earlier as the data set does not support this. Again, the one male patient is a clear outlier apart from the majority of the data.

To validate and better understand these findings, the results were compared with published research on the role of tau in Alzheimer’s disease. Iqbal et al. reported that tau pathology is strongly associated with Alzheimer’s disease and neurodegeneration. Importantly, abnormal hyperphosphorylation of tau, rather than simply the presence or total amount of tau, was identified as a major feature of Alzheimer’s pathology. Hyperphosphorylated tau loses its normal ability to stabilize microtubules and gains toxic functions that contribute to neuronal degeneration (16). Abnormally hyperphosphorylated tau also accumulates in neurofibrillary tangles and is associated with neurodegeneration and dementia (16). 

Medeiros et al. provided additional evidence that the relationship between tau and Alzheimer’s disease is more complex than simply having a higher total tau level. Under normal conditions, tau contributes to the stabilization of neuronal microtubules, while in Alzheimer’s disease, abnormal hyperphosphorylation reduces tau's interaction with microtubules and promotes its aggregation into neurofibrillary tangles (17). Multiple processes can also contribute to abnormal tau pathology, including changes in kinase and phosphatase activity, Aβ accumulation, impaired glucose metabolism, and inflammation (17). These findings suggest that Alzheimer’s disease progression involves multiple interacting biological mechanisms rather than tTAU concentration alone. 

These studies provide a possible explanation for the results of the present analysis despite the established importance of tau in Alzheimer’s disease. The present analysis measured tTAU, whereas the pathological effects described in the literature depend heavily on the form and modification of tau, particularly hyperphosphorylated and aggregated tau. Soluble phosphorylated tau has been associated with neuronal degeneration, while neurofibrillary tangle accumulation may occur somewhat independently of memory loss (17). Therefore, a higher overall tTAU level does not necessarily indicate a greater amount of the specific toxic forms of tau or an earlier age at death. 

Overall, the published literature supports the biological importance of tau while also demonstrating that tTAU alone may not capture the complexity of tau pathology. This is consistent with the finding that tTAU was not strongly associated with age at death and did not significantly differ between female and male patients in the present dataset. However, the two studies do not directly validate the sex-specific finding or the relationship between tTAU and age at death, as the primary focus of the literature is the biological role and pathological forms of tau. Additional studies specifically examining sex, tTAU levels, and mortality would be necessary for stronger external validation.
